```python
import pandas as pd

# Define the files uploaded in your Colab session
files = {
    'Accounts': 'cleaned_accounts.xlsx',
    'Invoices': 'cleaned_invoices.xlsx',
    'Subscriptions': 'cleaned_subscriptions.xlsx',
    'Support Tickets': 'cleaned_support_tickets.xlsx',
    'Users': 'cleaned_users.xlsx'
}

# Load each file into a dictionary of DataFrames
dataframes = {}
for name, filename in files.items():
    dataframes[name] = pd.read_excel(filename)
    print(f"Successfully loaded: {filename}")
```

```python
def perform_eda(df, name):
    print(f"==================================================")
    print(f"            Exploratory Data Analysis: {name}")
    print(f"==================================================\n")

    print("1. Shape of the dataset (Rows, Columns):")
    print(df.shape, "\n")

    print("2. First 5 rows:")
    display(df.head())

    print("\n3. Data Types and Non-Null Counts:")
    df.info()

    print("\n4. Summary Statistics:")
    display(df.describe(include='all'))

    print("\n5. Missing Values:")
    missing_vals = df.isnull().sum()
    print(missing_vals[missing_vals > 0] if missing_vals.sum() > 0 else "No missing values.")
    print("\n\n")
```

```python
# Run EDA for all loaded tables sequentially
for name, df in dataframes.items():
    perform_eda(df, name)
```

```python
# Change 'Users' to 'Accounts', 'Invoices', 'Subscriptions', or 'Support Tickets'
table_to_analyze = 'Users'

perform_eda(dataframes[table_to_analyze], table_to_analyze)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set a clean visual theme for all plots
sns.set_theme(style="whitegrid", palette="muted")
print("Visualization libraries loaded and theme set.")
```

```python
accounts_df = dataframes['Accounts']

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 1. Accounts by Industry
sns.countplot(data=accounts_df, y='industry', order=accounts_df['industry'].value_counts().index, ax=axes[0])
axes[0].set_title('Number of Accounts by Industry')
axes[0].set_xlabel('Count')
axes[0].set_ylabel('Industry')

# 2. Accounts by Company Size
size_counts = accounts_df['company_size'].value_counts()
axes[1].pie(size_counts, labels=size_counts.index, autopct='%1.1f%%', startangle=90)
axes[1].set_title('Distribution of Company Size')

plt.tight_layout()
plt.show()
```

```python
invoices_df = dataframes['Invoices']
subs_df = dataframes['Subscriptions']

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 1. Invoice Status Distribution
sns.countplot(data=invoices_df, x='status', order=invoices_df['status'].value_counts().index, ax=axes[0])
axes[0].set_title('Invoice Status Distribution')
axes[0].set_xlabel('Status')
axes[0].set_ylabel('Count')

# 2. Distribution of MRR (Monthly Recurring Revenue)
sns.histplot(data=subs_df, x='mrr', bins=30, kde=True, ax=axes[1], color='green')
axes[1].set_title('Distribution of Monthly Recurring Revenue (MRR)')
axes[1].set_xlabel('MRR ($)')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
tickets_df = dataframes['Support Tickets']

fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# 1. Tickets by Category
sns.countplot(data=tickets_df, y='category', order=tickets_df['category'].value_counts().index, ax=axes[0])
axes[0].set_title('Support Tickets by Category')
axes[0].set_xlabel('Count')
axes[0].set_ylabel('Category')

# 2. Resolution Hours by Priority (Excluding Open Tickets)
resolved_tickets = tickets_df.dropna(subset=['resolution_hours'])
sns.boxplot(data=resolved_tickets, x='priority', y='resolution_hours',
            order=['low', 'normal', 'high', 'urgent'], ax=axes[1])
axes[1].set_title('Resolution Hours by Priority Level')
axes[1].set_xlabel('Priority')
axes[1].set_ylabel('Resolution Time (Hours)')

plt.tight_layout()
plt.show()
```
